## Logic Building Task – 1

    1. User Login Check
    Problem Statement:
    Given a username and password, check whether login is successful.
    username = "admin"
    password = "1234"
    
    Requirements:
    Print "Login Successful" if both username and password match
    Otherwise print "Invalid Credentials"
    Real-World Application: Authentication systems


```python
username = input("username= ").strip().lower()
password = input("password= ").strip()

if username == "admin" and password == "1234":
    print("Login Successful")
else:
    print("Invalid Credentials")
```

    2. Pass / Fail Analyzer
    Problem Statement:
    Analyze student results.
    marks = [45, 78, 90, 33, 60]
    
    Requirements:
    A student passes if marks ≥ 50
    Count the total number of pass students
    Count the total number of fail students
    Print the final result clearly
    Real-World Application: Academic evaluation systems


```python
marks = [45, 78, 90, 33, 60]

passed = sum(1 for m in marks if m >= 50)
failed = len(marks) - passed

print(f"Passed: {passed}, Failed: {failed}")
```

    3. Simple Data Cleaner
    Problem Statement:
    Clean and standardize user names.
    names = [" Alice ", "bob", " CHARLIE "]
    
    Requirements:
    Remove extra spaces
    Convert all names to lowercase
    Print the cleaned list
    Real-World Application: Data preprocessing before analysis


```python
names = [" Alice ", "bob", " CHARLIE "]
cleaned = [name.strip().lower() for name in names]
print(cleaned)
```

    4. Message Length Analyzer
    Problem Statement:
    Analyze message sizes.
    messages = ["Hi", "Welcome to the platform", "OK"]
    
    Requirements:
    Print the length of each message
    Flag messages longer than 10 characters
    Real-World Application: Text filtering and validation systems


```python
messages = ["Hi", "Welcome to the platform", "OK"]

for i in messages:
    length = len(i)
    flag = "(Longer Than 10 characters)" if length > 10 else ""
    print(f"{i}: {length} {flag}")
```

    5. Error Message Detector
    Problem Statement:
    Detect error messages from system logs.
    logs = ["INFO", "ERROR", "WARNING", "ERROR"]
    
    Requirements:
    Count the number of "ERROR" entries


```python
errors = sum(1 for log in logs if log.upper() == "ERROR")
print(f"errors: {errors}")
```
